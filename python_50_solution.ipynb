```python
def factorial(n: int) -> int:
    """
    Calculate the factorial of a non-negative integer.

    Args:
        n (int): The number to calculate factorial for.

    Returns:
        int: Factorial of the given number.

    Raises:
        TypeError: If input is not an integer.
        ValueError: If input is negative.
    """
    if not isinstance(n, int):
        raise TypeError("Input must be an integer")
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers")

    result = 1
    for i in range(1, n + 1):
        result *= i
    return result


if __name__ == "__main__":
    try:
        number = int(input("Enter a non-negative integer: "))
        print(f"Factorial of {number} = {factorial(number)}")
    except TypeError as te:
        print(te)
    except ValueError as ve:
        print(ve)
    except Exception as e:
        print(e)
```

```python
def second_largest(numbers: list) -> int:
    """
    Find the second largest element in a list.

    Args:
        numbers (list): A list of integers.

    Returns:
        int: The second largest element.

    Raises:
        TypeError: If input is not a list.
        ValueError: If list has fewer than 2 unique elements.
    """
    if not isinstance(numbers, list):
        raise TypeError("Input must be a list")
    if len(set(numbers)) < 2:
        raise ValueError("List must contain at least two unique elements")

    largest = second = float('-inf')
    for num in numbers:
        if num > largest:
            second = largest
            largest = num
        elif num > second and num != largest:
            second = num
    return second


if __name__ == "__main__":
    try:
        nums = [12, 45, 67, 23, 89, 34, 89]  
        print("Second largest element =", second_largest(nums))
    except TypeError as te:
        print(te)
    except ValueError as ve:
        print(ve)
    except Exception as e:
        print(e)
```

Factorial Using Recursion

Write a recursive function to calculate the factorial of a given number

```python
def factorial(n: int) -> int:
    """
    Calculate factorial of a non-negative integer using recursion.

    Args:
        n (int): The number to calculate factorial for.

    Returns:
        int: Factorial of the given number.

    Raises:
        TypeError: If input is not an integer.
        ValueError: If input is negative.
    """
    if not isinstance(n, int):
        raise TypeError("Input must be an integer")
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers")

    # Base 
    if n == 0 or n == 1:
        return 1
    # Recursive case
    return n * factorial(n - 1)
if __name__ == "__main__":
    try:
        number = int(input("Enter a non-negative integer: "))
        print(f"Factorial of {number} = {factorial(number)}")
    except TypeError as te:
        print(te)
    except ValueError as ve:
        print(ve)
    except Exception as e:
        print(e)
```

Count Occurrences of Each Character in a String

Take a string as input and count the occurrences of each character

```python
def count_characters(s: str) -> dict:
    """
    Count occurrences of each character in a string.

    Args:
        s (str): Input string.

    Returns:
        dict: Dictionary with characters as keys and their counts as values.

    Raises:
        TypeError: If input is not a string.
        ValueError: If input is empty.
    """
    if not isinstance(s, str):
        raise TypeError("Input must be a string")
    if len(s) == 0:
        raise ValueError("String cannot be empty")

    char_count = {}
    for char in s:
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1
    return char_count


if __name__ == "__main__":
    try:
        text = input("Enter a string: ")
        result = count_characters(text)
        print("Character occurrences:")
        for char, count in result.items():
            print(f"'{char}': {count}")
    except TypeError as te:
        print(te)
    except ValueError as ve:
        print(ve)
    except Exception as e:
        print(e)
```

## 13.

Check for Anagram

Write a function to check if two input strings are anagrams of each other.

```python
def is_anagram(str1: str, str2: str) -> bool:
    """
    Check if two strings are anagrams of each other.

    Args:
        str1 (str): First string.
        str2 (str): Second string.

    Returns:
        bool: True if strings are anagrams, False otherwise.

    Raises:
        TypeError: If inputs are not strings.
        ValueError: If any string is empty.
    """
    if not isinstance(str1, str) or not isinstance(str2, str):
        raise TypeError("Inputs must be strings")
    if len(str1) == 0 or len(str2) == 0:
        raise ValueError("Strings cannot be empty")

    # Normalize case and remove spaces
    str1 = str1.replace(" ", "").lower()
    str2 = str2.replace(" ", "").lower()

    # Compare character counts using for loop
    if len(str1) != len(str2):
        return False

    char_count = {}
    for ch in str1:
        char_count[ch] = char_count.get(ch, 0) + 1

    for ch in str2:
        if ch not in char_count:
            return False
        char_count[ch] -= 1
        if char_count[ch] < 0:
            return False

    return True


if __name__ == "__main__":
    try:
        s1 = input("Enter first string: ")
        s2 = input("Enter second string: ")
        if is_anagram(s1, s2):
            print(f"'{s1}' and '{s2}' are anagrams")
        else:
            print(f"'{s1}' and '{s2}' are not anagrams")
    except TypeError as te:
        print(te)
    except ValueError as ve:
        print(ve)
    except Exception as e:
        print(e)
```

## 14
Fibonacci Sequence (Iterative)

Write a function that prints the first n numbers in the Fibonacci sequence

```python
def fibonacci(n: int) -> list:
    """
    Generate the first n numbers in the Fibonacci sequence (iterative).

    Args:
        n (int): Number of terms to generate.

    Returns:
        list: List containing the first n Fibonacci numbers.

    Raises:
        TypeError: If input is not an integer.
        ValueError: If input is less than 1.
    """
    if not isinstance(n, int):
        raise TypeError("Input must be an integer")
    if n < 1:
        raise ValueError("Number of terms must be at least 1")

    sequence = []
    a, b = 0, 1
    for _ in range(n):
        sequence.append(a)
        a, b = b, a + b
    return sequence


if __name__ == "__main__":
    try:
        terms = int(input("Enter how many Fibonacci numbers to print: "))
        print(f"First {terms} Fibonacci numbers:")
        print(fibonacci(terms))
    except TypeError as te:
        print(te)
    except ValueError as ve:
        print(ve)
    except Exception as e:
        print(e)
```

Calculate Power without Using ** Operator

Write a function that calculates xy  (x raised to the power of y) without using Python’s power operator (**).

```python
def power(x: int, y: int) -> int:
    """
    Calculate x raised to the power y without using ** operator.

    Args:
        x (int): Base number.
        y (int): Exponent (non-negative integer).

    Returns:
        int: Result of x^y.

    Raises:
        TypeError: If inputs are not integers.
        ValueError: If exponent is negative.
    """
    if not isinstance(x, int) or not isinstance(y, int):
        raise TypeError("Both inputs must be integers")
    if y < 0:
        raise ValueError("Exponent must be a non-negative integer")

    result = 1
    for _ in range(y):
        result *= x
    return result


if __name__ == "__main__":
    try:
        base = int(input("Enter base (x): "))
        exponent = int(input("Enter exponent (y): "))
        print(f"{base}^{exponent} = {power(base, exponent)}")
    except TypeError as te:
        print(te)
    except ValueError as ve:
        print(ve)
    except Exception as e:
        print(e)
```

## 18
Binary Search Implementation

Write a function that performs binary search on a sorted list and returns the index of the target element or -1 if not found

```python
def binary_search(arr: list, target: int) -> int:
    """
    Perform binary search on a sorted list.

    Args:
        arr (list): A sorted list of integers.
        target (int): The element to search for.

    Returns:
        int: Index of target element if found, else -1.

    Raises:
        TypeError: If arr is not a list or target is not an integer.
        ValueError: If list is empty.
    """
    if not isinstance(arr, list):
        raise TypeError("Input must be a list")
    if not isinstance(target, int):
        raise TypeError("Target must be an integer")
    if len(arr) == 0:
        raise ValueError("List cannot be empty")

    low, high = 0, len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1

    return -1


if __name__ == "__main__":
    try:
        sorted_list = [10, 20, 30, 40, 50, 60, 70]
        target = int(input("Enter the number to search: "))
        result = binary_search(sorted_list, target)
        if result != -1:
            print(f"{target} found at index {result}")
        else:
            print(f"{target} not found in the list")
    except TypeError as te:
        print(te)
    except ValueError as ve:
        print(ve)
    except Exception as e:
        print(e)
```

## 19
Reverse Words in a Sentence

Note: Don’t use split or join.

Write a function to reverse the order of words in a sentence, keeping the words themselves intact (sentence should not be 
changed or altered in any way; they should remain exactly as they are,)

```python
def reverse_words(sentence: str) -> str:
    """
    Reverse the order of words in a sentence without using split or join.

    Args:
        sentence (str): Input sentence.

    Returns:
        str: Sentence with words in reversed order.

    Raises:
        TypeError: If input is not a string.
        ValueError: If input is empty.
    """
    if not isinstance(sentence, str):
        raise TypeError("Input must be a string")
    if len(sentence.strip()) == 0:
        raise ValueError("Sentence cannot be empty")
    words = []
    word = ""
    result = ""
    for char in sentence:
        if char == " ":
            if word:
                words.append(word)
                word = ""
        else:
            word += char
    if word:  # add last word
        words.append(word)

    # Reverse order of words manually
    for i in range(len(words) - 1, -1, -1):
        result += words[i]
        if i != 0:
            result += " "

    return result


if __name__ == "__main__":
    try:
        text = input("Enter a sentence: ")
        print("Reversed sentence:", reverse_words(text))
    except TypeError as te:
        print(te)
    except ValueError as ve:
        print(ve)
    except Exception as e:
        print(e)
```

## 20
Calculate GCD.

Write a function to find the Greatest Common Divisor (GCD).

```python
def gcd(a: int, b: int) -> int:
    """
    Calculate the Greatest Common Divisor (GCD) of two integers using Euclidean Algorithm.

    Args:
        a (int): First integer.
        b (int): Second integer.

    Returns:
        int: GCD of a and b.

    Raises:
        TypeError: If inputs are not integers.
        ValueError: If both inputs are zero.
    """
    if not isinstance(a, int) or not isinstance(b, int):
        raise TypeError("Inputs must be integers")
    if a == 0 and b == 0:
        raise ValueError("GCD is not defined for both numbers being zero")

    a, b = abs(a), abs(b)  
    while b != 0:
        a, b = b, a % b
    return a
if __name__ == "__main__":
    try:
        x = int(input("Enter first number: "))
        y = int(input("Enter second number: "))
        print(f"GCD of {x} and {y} = {gcd(x, y)}")
    except TypeError as te:
        print(te)
    except ValueError as ve:
        print(ve)
    except Exception as e:
        print(e)
```

## 22
Convert Decimal to Binary (Without using bin () function)

Write a function to convert a given decimal number to its binary representation

```python
def decimal_to_binary(n: int) -> str:
    """
    Convert a decimal number to its binary representation without using bin().

    Args:
        n (int): Decimal number.

    Returns:
        str: Binary representation of the number.

    Raises:
        TypeError: If input is not an integer.
        ValueError: If input is negative.
    """
    if not isinstance(n, int):
        raise TypeError("Input must be an integer")
    if n < 0:
        raise ValueError("Negative numbers are not supported")

    if n == 0:
        return "0"

    binary = ""
    while n > 0:
        remainder = n % 2
        binary = str(remainder) + binary
        n //= 2
    return binary
if __name__ == "__main__":
    try:
        number = int(input("Enter a decimal number: "))
        print(f"Binary representation of {number} = {decimal_to_binary(number)}")
    except TypeError as te:
        print(te)
    except ValueError as ve:
        print(ve)
    except Exception as e:
        print(e)
```

## 23. Find Intersection of Two Lists
 Write a function that returns the intersection of two lists without using set operations

```python
def list_intersection(list1, list2):
    """
    Function to find intersection of two lists
    without using set operations.
    """
    intersection = []
    for item in list1:
        if item in list2 and item not in intersection:
            intersection.append(item)
    return intersection


if __name__ == "__main__":
    try:
        listA = [int(x) for x in input("Enter first list (space separated): ").split()]
        listB = [int(x) for x in input("Enter second list (space separated): ").split()]

        
        result = list_intersection(listA, listB)
        print("Intersection of lists:", result)

    except ValueError:
        print("Error: Please enter only integers separated by spaces.")
    except Exception as e:
        print(" Unexpected Error:", e)
```

## 24. Remove All Whitespace from a String
Write a function that removes all whitespace characters (spaces, tabs, newlines) from a given string.

```python
def remove_whitespace(s: str) -> str:
    """
    Remove all whitespace characters (spaces, tabs, newlines) from a string.

    Args:
        s (str): Input string.

    Returns:
        str: String without any whitespace characters.

    Raises:
        TypeError: If input is not a string.
        ValueError: If input is empty.
    """
    if not isinstance(s, str):
        raise TypeError("Input must be a string")
    if len(s) == 0:
        raise ValueError("String cannot be empty")

    result = ""
    for char in s:
        if not char.isspace():   
            result += char
    return result


if __name__ == "__main__":
    try:
        text = input("Enter a string: ")
        print("String without whitespace:", remove_whitespace(text))
    except TypeError as te:
        print(te)
    except ValueError as ve:
        print(ve)
    except Exception as e:
        print(e)
```

```python
def sum_list_recursive(numbers: list) -> int:
    """
    Calculate the sum of a list of numbers using recursion.

    Args:
        numbers (list): List of integers.

    Returns:
        int: Sum of all numbers in the list.

    Raises:
        TypeError: If input is not a list or contains non-integers.
        ValueError: If list is empty.
    """
    if not isinstance(numbers, list):
        raise TypeError("Input must be a list")
    if len(numbers) == 0:
        raise ValueError("List cannot be empty")
    if not all(isinstance(num, (int, float)) for num in numbers):
        raise TypeError("List must contain only numbers")

    # Base case
    if len(numbers) == 1:
        return numbers[0]
    # Recursive case
    return numbers[0] + sum_list_recursive(numbers[1:])


if __name__ == "__main__":
    try:
        nums = [10, 20, 30, 40, 50]
        print("List:", nums)
        print("Sum of list =", sum_list_recursive(nums))
    except TypeError as te:
        print(te)
    except ValueError as ve:
        print(ve)
    except Exception as e:
        print(e)
```

## 27. Merge Two Sorted Lists
Write a function to merge two sorted lists into a single sorted list.

```python
def merge_sorted_lists(list1: list, list2: list) -> list:
    """
    This function merges two sorted lists into a single sorted list.

    Args:
        list1: Accepts the first sorted list as argument and it should be a list.
        list2: Accepts the second sorted list as argument and it should be a list.

    Returns:
        merged_list: Returns a merged sorted list.

    Raises:
        TypeError: If either input is not a list.
    """
    if not isinstance(list1, list):
        raise TypeError("First input should be a list only")

    if not isinstance(list2, list):
        raise TypeError("Second input should be a list only")

    merged_list = []
    i = 0
    j = 0

    while i < len(list1) and j < len(list2):
        if list1[i] <= list2[j]:
            merged_list.append(list1[i])
            i += 1
        else:
            merged_list.append(list2[j])
            j += 1

    while i < len(list1):
        merged_list.append(list1[i])
        i += 1

    while j < len(list2):
        merged_list.append(list2[j])
        j += 1

    return merged_list


if __name__ == "__main__":
    try:
        list1 = list(map(int, input("Enter elements of first sorted list separated by spaces: ").split()))
        list2 = list(map(int, input("Enter elements of second sorted list separated by spaces: ").split()))

        print("Merged Sorted List:", merge_sorted_lists(list1, list2))

    except TypeError as te:
        print(te)

    except ValueError as ve:
        print(ve)

    except Exception as e:
        print(e)
```

## 32. Reverse Integer
Write a function that takes an integer and returns its digits reversed

```python
def reverse_integer(number: int) -> int:
    """
    This function reverses an integer.

    Args:
        number: Accepts an integer as argument and it should be of type int.

    Returns:
        reversed_num: Returns the reversed integer.

    Raises:
        TypeError: If the input is not an integer.
    """
    if not isinstance(number, int):
        raise TypeError("Your input should be integer only")

    sign = -1 if number < 0 else 1
    number = abs(number)

    reversed_num = 0

    while number > 0:
        digit = number % 10
        reversed_num = reversed_num * 10 + digit
        number //= 10

    return sign * reversed_num


if __name__ == "__main__":
    try:
        number = int(input("Enter an integer: "))
        print(reverse_integer(number))

    except TypeError as te:
        print(te)

    except ValueError as ve:
        print(ve)

    except Exception as e:
        print(e)
```

## 33. Implement a Stack using List
Create a stack with push, pop, and peek operations using a list.

Stack: A stack is a linear data structure that follows the Last In, First Out (LIFO) principle. This means that the last element 
added to the stack is the first one to be removed. Think of it like a stack of plates; you can only add or remove plates from the 
top.

List: A list is a more general data structure that can store a collection of items. It allows for random access and can be used to 
store elements in any order. Lists do not have strict rules about how elements are added or removed

```python
class Stack:
    """
    This class implements a Stack using a list.

    Stack follows the Last In, First Out (LIFO) principle.
    """

    def __init__(self) -> None:
        """
        Initializes an empty stack.
        """
        self.stack = []

    def push(self, item) -> None:
        """
        Adds an element to the top of the stack.

        Args:
            item: Element to be added to the stack.

        Returns:
            None
        """
        self.stack.append(item)

    def pop(self):
        """
        Removes and returns the top element of the stack.

        Returns:
            item: The top element of the stack.

        Raises:
            IndexError: If the stack is empty.
        """
        if not self.stack:
            raise IndexError("Stack is empty. Cannot pop element.")

        return self.stack.pop()

    def peek(self):
        """
        Returns the top element of the stack without removing it.

        Returns:
            item: The top element of the stack.

        Raises:
            IndexError: If the stack is empty.
        """
        if not self.stack:
            raise IndexError("Stack is empty. Cannot peek element.")

        return self.stack[-1]

    def display(self) -> list:
        """
        Displays all elements of the stack.

        Returns:
            list: Current stack elements.
        """
        return self.stack


if __name__ == "__main__":
    try:
        stack = Stack()

        stack.push(10)
        stack.push(20)
        stack.push(30)

        print("Stack:", stack.display())
        print("Top Element:", stack.peek())

        print("Popped Element:", stack.pop())
        print("Stack After Pop:", stack.display())

    except IndexError as ie:
        print(ie)

    except Exception as e:
        print(e)
```

```python
def check_power_of_two(number: int) -> tuple:
    """
    This function checks whether a number is a power of 2
    and returns the corresponding power.

    Args:
        number: Accepts a number as argument and it should be an integer.

    Returns:
        tuple:
            (True, power) if the number is a power of 2.
            (False, None) if the number is not a power of 2.

    Raises:
        TypeError: If the input is not an integer.
        ValueError: If the input is less than or equal to 0.
    """
    if not isinstance(number, int):
        raise TypeError("Your input should be an integer only")

    if number <= 0:
        raise ValueError("Number should be greater than 0")

    power = 0
    temp = number

    while temp > 1:
        if temp % 2 != 0:
            return False, None

        temp = temp // 2
        power += 1

    return True, power


if __name__ == "__main__":
    try:
        number = int(input("Enter a number: "))

        is_power, power = check_power_of_two(number)

        if is_power:
            print(f"{number} is a power of 2")
            print(f"{number} = 2^{power}")
        else:
            print(f"{number} is not a power of 2")

    except TypeError as te:
        print(te)

    except ValueError as ve:
        print(ve)

    except Exception as e:
        print(e)
```

## 39. Implement Queue using List
Create a queue with enqueue, dequeue, and peek operations using a list

```python
class Queue:
    """
    This class implements a Queue using a list.

    Queue follows the First In, First Out (FIFO) principle.
    """

    def __init__(self) -> None:
        """
        Initializes an empty queue.
        """
        self.queue = []

    def enqueue(self, item) -> None:
        """
        Adds an element to the rear of the queue.

        Args:
            item: Element to be added to the queue.

        Returns:
            None
        """
        self.queue.append(item)

    def dequeue(self):
        """
        Removes and returns the front element of the queue.

        Returns:
            item: The front element of the queue.

        Raises:
            IndexError: If the queue is empty.
        """
        if len(self.queue) == 0:
            raise IndexError("Queue is empty. Cannot dequeue element.")

        return self.queue.pop(0)

    def peek(self):
        """
        Returns the front element of the queue without removing it.

        Returns:
            item: The front element of the queue.

        Raises:
            IndexError: If the queue is empty.
        """
        if len(self.queue) == 0:
            raise IndexError("Queue is empty. Cannot peek element.")

        return self.queue[0]

    def display(self) -> list:
        """
        Displays all elements of the queue.

        Returns:
            list: Current queue elements.
        """
        return self.queue


if __name__ == "__main__":
    try:
        q = Queue()

        q.enqueue(10)
        q.enqueue(20)
        q.enqueue(30)

        print("Queue:", q.display())

        print("Front Element:", q.peek())

        print("Dequeued Element:", q.dequeue())

        print("Queue After Dequeue:", q.display())

    except IndexError as ie:
        print(ie)

    except Exception as e:
        print(e)
```

## 40. Find the Median of a List of Numbers
Write a function to calculate the median of a list of numbers

```python
def find_median(numbers: list) -> float:
    """
    This function calculates the median of a list of numbers.

    Args:
        numbers: Accepts a list of numbers as argument.

    Returns:
        median: Returns the median of the list.

    Raises:
        TypeError: If the input is not a list.
        ValueError: If the list is empty.
    """
    if not isinstance(numbers, list):
        raise TypeError("Your input should be a list only")

    if len(numbers) == 0:
        raise ValueError("List cannot be empty")

    sorted_numbers = sorted(numbers)
    n = len(sorted_numbers)

    if n % 2 == 1:
        median = sorted_numbers[n // 2]
    else:
        median = (
            sorted_numbers[n // 2 - 1] +
            sorted_numbers[n // 2]
        ) / 2

    return median


if __name__ == "__main__":
    try:
        numbers = list(
            map(
                float,
                input(
                    "Enter numbers separated by spaces: "
                ).split()
            )
        )

        print("Median:", find_median(numbers))

    except TypeError as te:
        print(te)

    except ValueError as ve:
        print(ve)

    except Exception as e:
        print(e)
```

## 42. Find LCM of two numbers

```python
def find_lcm(num1: int, num2: int) -> int:
    """
    This function finds the Least Common Multiple (LCM)
    of two numbers.

    Args:
        num1: Accepts the first number as an integer.
        num2: Accepts the second number as an integer.

    Returns:
        lcm: Returns the LCM of the two numbers.

    Raises:
        TypeError: If any input is not an integer.
        ValueError: If any input is less than or equal to 0.
    """
    if not isinstance(num1, int):
        raise TypeError("First input should be an integer only")

    if not isinstance(num2, int):
        raise TypeError("Second input should be an integer only")

    if num1 <= 0 or num2 <= 0:
        raise ValueError("Both numbers should be greater than 0")

    greater = max(num1, num2)

    while True:
        if greater % num1 == 0 and greater % num2 == 0:
            lcm = greater
            break

        greater += 1

    return lcm


if __name__ == "__main__":
    try:
        num1 = int(input("Enter first number: "))
        num2 = int(input("Enter second number: "))

        print("LCM:", find_lcm(num1, num2))

    except TypeError as te:
        print(te)

    except ValueError as ve:
        print(ve)

    except Exception as e:
        print(e)
```

## 43. Convert Binary to Decimal
Write a function to convert a binary number (as a string) to a decimal


```python
def binary_to_decimal(binary: str) -> int:
    """
    This function converts a binary number (as a string)
    to its decimal equivalent.

    Args:
        binary: Accepts a binary number as a string.

    Returns:
        decimal: Returns the decimal equivalent of the binary number.

    Raises:
        TypeError: If the input is not a string.
        ValueError: If the string contains characters other than 0 and 1.
    """
    if not isinstance(binary, str):
        raise TypeError("Your input should be a string only")

    for digit in binary:
        if digit not in ("0", "1"):
            raise ValueError("Binary number should contain only 0 and 1")

    decimal = 0
    power = 0

    for digit in reversed(binary):
        decimal += int(digit) * (2 ** power)
        power += 1

    return decimal


if __name__ == "__main__":
    try:
        binary = input("Enter a binary number: ")

        print("Decimal Number:", binary_to_decimal(binary))

    except TypeError as te:
        print(te)

    except ValueError as ve:
        print(ve)

    except Exception as e:
        print(e)
```

## 46. Remove Nth Occurrence of a Character from a String
Write a function that removes the nth occurrence of a character from a string


```python
def remove_nth_occurrence(text: str, char: str, n: int) -> str:
    """
    This function removes the nth occurrence of a character
    from a string.

    Args:
        text: Accepts a string.
        char: Accepts a character to be removed.
        n: Accepts the occurrence number to remove.

    Returns:
        result: Returns the modified string after removing
                the nth occurrence of the character.

    Raises:
        TypeError: If input types are invalid.
        ValueError: If n is less than or equal to 0.
    """
    if not isinstance(text, str):
        raise TypeError("First input should be a string only")

    if not isinstance(char, str):
        raise TypeError("Second input should be a string only")

    if not isinstance(n, int):
        raise TypeError("Third input should be an integer only")

    if len(char) != 1:
        raise ValueError("Character input should contain exactly one character")

    if n <= 0:
        raise ValueError("Occurrence number should be greater than 0")

    count = 0
    result = ""

    for c in text:
        if c == char:
            count += 1

            if count == n:
                continue

        result += c

    return result


if __name__ == "__main__":
    try:
        text = input("Enter a string: ")
        char = input("Enter the character to remove: ")
        n = int(input("Enter occurrence number to remove: "))

        print("Modified String:", remove_nth_occurrence(text, char, n))

    except TypeError as te:
        print(te)

    except ValueError as ve:
        print(ve)

    except Exception as e:
        print(e)
```

## 47. Find the First Non-Repeated Character in a String
Write a function to return the first non-repeated character in a string


```python
def first_non_repeated_character(text: str) -> str:
    """
    This function finds and returns the first non-repeated
    character in a string.

    Args:
        text: Accepts a string as argument.

    Returns:
        char: Returns the first non-repeated character.

    Raises:
        TypeError: If the input is not a string.
        ValueError: If no non-repeated character is found.
    """
    if not isinstance(text, str):
        raise TypeError("Your input should be a string only")

    frequency = {}

    for char in text:
        frequency[char] = frequency.get(char, 0) + 1

    for char in text:
        if frequency[char] == 1:
            return char

    raise ValueError("No non-repeated character found")


if __name__ == "__main__":
    try:
        text = input("Enter a string: ")

        print(
            "First Non-Repeated Character:",
            first_non_repeated_character(text)
        )

    except TypeError as te:
        print(te)

    except ValueError as ve:
        print(ve)

    except Exception as e:
        print(e)
```

## 48. Check if String Contains Only Digits
Write a function that checks if a string contains only numeric characters


```python
def contains_only_digits(text: str) -> bool:
    """
    This function checks whether a string contains
    only numeric characters.

    Args:
        text: Accepts a string as argument.

    Returns:
        True: If the string contains only digits.
        False: If the string contains non-numeric characters.

    Raises:
        TypeError: If the input is not a string.
        ValueError: If the string is empty.
    """
    if not isinstance(text, str):
        raise TypeError("Your input should be a string only")

    if len(text) == 0:
        raise ValueError("String cannot be empty")

    for char in text:
        if char < '0' or char > '9':
            return False

    return True


if __name__ == "__main__":
    try:
        text = input("Enter a string: ")

        if contains_only_digits(text):
            print("The string contains only digits.")
        else:
            print("The string does not contain only digits.")

    except TypeError as te:
        print(te)

    except ValueError as ve:
        print(ve)

    except Exception as e:
        print(e)
```

## 49. Check if Number is Armstrong Number
Write a function that checks if a number is an Armstrong number


```python
def is_armstrong_number(number: int) -> bool:
    """
    This function checks whether a number is an
    Armstrong number or not.

    Args:
        number: Accepts a number as an integer.

    Returns:
        True: If the number is an Armstrong number.
        False: If the number is not an Armstrong number.

    Raises:
        TypeError: If the input is not an integer.
        ValueError: If the number is negative.
    """
    if not isinstance(number, int):
        raise TypeError("Your input should be an integer only")

    if number < 0:
        raise ValueError("Number should not be negative")

    num_str = str(number)
    power = len(num_str)

    total = 0

    for digit in num_str:
        total += int(digit) ** power

    return total == number


if __name__ == "__main__":
    try:
        number = int(input("Enter a number: "))

        if is_armstrong_number(number):
            print(f"{number} is an Armstrong Number")
        else:
            print(f"{number} is not an Armstrong Number")

    except TypeError as te:
        print(te)

    except ValueError as ve:
        print(ve)

    except Exception as e:
        print(e)
```

## 50. Reverse an Array (List) 
The function should not return a new array; 
it should modify the original input List do not create new list.
The solution should run in O(n) time complexity.
The space complexity should be O(1) (constant extra space). 

```python
def reverse_list(numbers: list) -> None:
    """
    This function reverses a list in place without
    creating a new list.

    Args:
        numbers: Accepts a list as argument.

    Returns:
        None: The original list is modified directly.

    Raises:
        TypeError: If the input is not a list.
    """
    if not isinstance(numbers, list):
        raise TypeError("Your input should be a list only")

    left = 0
    right = len(numbers) - 1

    while left < right:
        numbers[left], numbers[right] = numbers[right], numbers[left]

        left += 1
        right -= 1


if __name__ == "__main__":
    try:
        numbers = list(
            map(
                int,
                input("Enter elements of the list separated by spaces: ").split()
            )
        )

        print("Original List:", numbers)

        reverse_list(numbers)

        print("Reversed List:", numbers)

    except TypeError as te:
        print(te)

    except ValueError as ve:
        print(ve)

    except Exception as e:
        print(e)
```

## 51. Check if Subsequence Exists in List
Write a function to check if a smaller list is a subsequence of a larger list


```python
def is_subsequence(main_list: list, sub_list: list) -> bool:
    """
    This function checks whether a smaller list is a
    subsequence of a larger list.

    Args:
        main_list: Accepts the larger list.
        sub_list: Accepts the smaller list.

    Returns:
        True: If sub_list is a subsequence of main_list.
        False: Otherwise.

    Raises:
        TypeError: If any input is not a list.
    """
    if not isinstance(main_list, list):
        raise TypeError("First input should be a list only")

    if not isinstance(sub_list, list):
        raise TypeError("Second input should be a list only")

    i = 0
    j = 0

    while i < len(main_list) and j < len(sub_list):
        if main_list[i] == sub_list[j]:
            j += 1
        i += 1

    return j == len(sub_list)


if __name__ == "__main__":
    try:
        main_list = list(
            map(
                int,
                input(
                    "Enter elements of the main list separated by spaces: "
                ).split()
            )
        )

        sub_list = list(
            map(
                int,
                input(
                    "Enter elements of the subsequence list separated by spaces: "
                ).split()
            )
        )

        if is_subsequence(main_list, sub_list):
            print("The second list is a subsequence of the first list.")
        else:
            print("The second list is not a subsequence of the first list.")

    except TypeError as te:
        print(te)

    except ValueError as ve:
        print(ve)

    except Exception as e:
        print(e)
```

## 52. Generate All Subsets of a Set
Write a function to generate all possible subsets of a given set (list) of numbers


```python
def generate_subsets(numbers: list) -> list:
    """
    This function generates all possible subsets
    of a given list.

    Args:
        numbers: Accepts a list of numbers.

    Returns:
        subsets: Returns a list containing all possible subsets.

    Raises:
        TypeError: If the input is not a list.
    """
    if not isinstance(numbers, list):
        raise TypeError("Your input should be a list only")

    subsets = [[]]

    for num in numbers:
        new_subsets = []

        for subset in subsets:
            new_subsets.append(subset + [num])

        subsets.extend(new_subsets)

    return subsets


if __name__ == "__main__":
    try:
        numbers = list(
            map(
                int,
                input(
                    "Enter elements of the list separated by spaces: "
                ).split()
            )
        )

        print("All Possible Subsets:")

        result = generate_subsets(numbers)

        for subset in result:
            print(subset)

    except TypeError as te:
        print(te)

    except ValueError as ve:
        print(ve)

    except Exception as e:
        print(e)
```
