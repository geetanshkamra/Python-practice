## 28. string.maketrans(x[, y[, z]]) and string.translate(table)
create a rulebook table for translate() and It can replace characters, delete characters, or use a dictionary mapping.

```python
#Replace 'i' with 'a' and 's' with 't'
rulebook1 = str.maketrans("is", "at")
string79 = "This is jupyter"
result = string79.translate(rulebook1)
print(result)  

#Replace a→1, b→2, c→3 and delete 'x'
rulebook2 = str.maketrans("abc", "123", "x")
string80 = "abcdex"
result = string80.translate(rulebook2)
print(result)  


#Delete characters 'a', 'e', and 'i'
rulebook3 = str.maketrans("", "", "aei")
string81 = "hey, delhi"
result = string81.translate(rulebook3)
print(result)  

#Replace using dictionary mapping
rulebook4 = str.maketrans({"a": "@", "b": "8", "c": "("})
string82 = "abcde"
result = string82.translate(rulebook4)
print(result)  
```

.maketrans() → creates a translation table (mapping of characters). easily replace or remove map or char

.translate(table) → applies that mapping to the string.
